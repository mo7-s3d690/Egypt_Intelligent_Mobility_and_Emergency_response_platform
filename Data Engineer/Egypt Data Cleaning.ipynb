```python
import pandas as pd
import pyodbc

# إعداد الاتصال بقاعدة البيانات
SERVER_NAME = r'DESKTOP-2A0QBS7\SQLEXPRESS'
DATABASE_NAME = 'EgyptTrafficDB'
conn_str = f'DRIVER={{ODBC Driver 17 for SQL Server}};SERVER={SERVER_NAME};DATABASE={DATABASE_NAME};Trusted_Connection=yes;'

try:
    conn = pyodbc.connect(conn_str)
    print("✅ تم الاتصال بقاعدة البيانات.. جاري سحب الجداول الثلاثة...")

    # 1. سحب الجداول
    df_locations = pd.read_sql("SELECT * FROM dbo.Locations", conn)
    df_weather = pd.read_sql("SELECT * FROM dbo.WeatherData", conn)
    df_traffic = pd.read_sql("SELECT * FROM dbo.Traffic_Logs", conn)

    # 2. تظبيط الوقت عشان نربط الطقس بالزحمة في نفس الساعة
    df_traffic['RecordedAt'] = pd.to_datetime(df_traffic['RecordedAt'])
    df_traffic['Join_Hour'] = df_traffic['RecordedAt'].dt.floor('h') # تقريب لأقرب ساعة

    df_weather['Weather_DateHour'] = pd.to_datetime(df_weather['Weather_DateHour'])

    # 3. الدمج الأول: المرور مع الأماكن
    df_step1 = pd.merge(
        df_traffic, 
        df_locations[['LocationID', 'LocationName', 'City', 'Latitude', 'Longitude']], 
        on='LocationID', 
        how='inner'
    )

    # 4. الدمج التاني: الناتج مع الطقس
    df_final = pd.merge(
        df_step1,
        df_weather[['City_Name', 'Weather_DateHour', 'Temperature_C', 'Humidity_Pct', 'Wind_Speed_ms', 'Weather_Condition']],
        left_on=['City', 'Join_Hour'],
        right_on=['City_Name', 'Weather_DateHour'],
        how='left'
    )

    # 5. تنظيف الأعمدة الزيادة وحفظ الملف
    columns_to_drop = ['Join_Hour', 'City_Name', 'Weather_DateHour']
    df_final = df_final.drop(columns=columns_to_drop, errors='ignore')

    output_file = 'Locations_Traffic_Weather_Merged.csv'
    df_final.to_csv(output_file, index=False, encoding='utf-8-sig')
    
    conn.close()
    
    print(f"🚀 عاش جداً! تم دمج {len(df_final)} سجل بنجاح.")
    print(f"📁 الملف اتحفظ باسم: {output_file}")
    
    # عرض أول سطرين للتأكيد
    display(df_final.head(10))

except Exception as e:
    print(f"❌ حصل خطأ: {e}")
```
